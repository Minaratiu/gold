```python
import tensorflow as tf

# Load model LSTM đã được thành viên khác train
model = tf.keras.models.load_model("gold_price_lstm_model.h5")

# Kiểm tra model nhận và trả về dữ liệu như thế nào
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
```

```python
import numpy as np

# Đọc dữ liệu test
X_test = np.load("X_test.npy")
y_test = np.load("y_test.npy")

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# Cho model dự đoán 499 mẫu test
pred_scaled = model.predict(X_test, verbose=0)

print("pred_scaled:", pred_scaled.shape)
```

```python
import joblib

# Load scaler được sử dụng khi train model
scaler = joblib.load("scaler.pkl")

# Chuyển giá dự đoán từ dạng chuẩn hóa về giá thật
pred_real = scaler.inverse_transform(
    pred_scaled
)

# Chuyển giá thực tế từ dạng chuẩn hóa về giá thật
y_test_real = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)

print("pred_real:", pred_real.shape)
print("y_test_real:", y_test_real.shape)
```

```python
print("5 giá thật:")
print(y_test_real[:5])

print("\n5 giá dự đoán:")
print(pred_real[:5])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Tính sai số giữa giá thật và giá dự đoán
mae = mean_absolute_error(y_test_real, pred_real)

# Tính RMSE
rmse = np.sqrt(
    mean_squared_error(y_test_real, pred_real)
)

print("MAE :", mae)
print("RMSE:", rmse)
```

```python
import pandas as pd

# Tạo bảng so sánh giá thật và giá dự đoán
result_df = pd.DataFrame({
    "Giá thực tế": y_test_real.flatten(),
    "Giá dự đoán": pred_real.flatten()
})

# Tính sai số tuyệt đối
result_df["Sai số"] = abs(
    result_df["Giá thực tế"] - result_df["Giá dự đoán"]
)

# Hiển thị 10 dòng đầu
result_df.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Vẽ giá thực tế
plt.plot(
    y_test_real,
    label="Giá thực tế"
)

# Vẽ giá dự đoán
plt.plot(
    pred_real,
    label="Giá dự đoán"
)

plt.title("So sánh giá vàng thực tế và giá dự đoán")
plt.xlabel("Mẫu test")
plt.ylabel("Giá vàng")
plt.legend()
plt.grid(True)

plt.show()
```
